```python
# Cargar dataset desde HuggingFace
from datasets import load_dataset

ds = load_dataset("mikeldiez/kortxovision_dataset", trust_remote_code=True)
print("Splits:", list(ds.keys()))

# Número de imágenes en cada split
for split in ds.keys():
    print(f"{split}: {len(ds[split])} imágenes")
```

```python
# Mostrar una imagen aleatoria de cada split
import random
from IPython.display import display

for split in ["test", "validation"]:
    if split not in ds:
        continue
    split_ds = ds[split]
    idx = random.randint(0, len(split_ds) - 1)
    sample = split_ds[idx]
    img = sample.get("image") or sample.get("images", [None])[0]
    if img is None:
        # Buscar columna de tipo Image
        for k, v in sample.items():
            if hasattr(v, "size"):
                img = v
                break
    if img is not None:
        display(img)
        print(f"--- {split} (idx={idx}) ---")
```
